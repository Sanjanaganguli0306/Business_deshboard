```python

import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px

# 1. PAGE CONFIGURATION
st.set_page_config(
    page_title="Executive Business Analytics Dashboard",
    page_icon="📊",
    layout="wide",
    initial_sidebar_state="expanded"
)

# Custom CSS for dark-themed cards and professional typography
st.markdown("""
    <style>
    .kpi-card {
        background-color: #1e1e24;
        padding: 20px;
        border-radius: 8px;
        box-shadow: 0 4px 6px rgba(0,0,0,0.3);
        text-align: center;
        border-left: 5px solid #00adb5;
    }
    .kpi-title {
        font-size: 14px;
        color: #a3a3a3;
        text-transform: uppercase;
        letter-spacing: 1px;
        margin-bottom: 10px;
    }
    .kpi-value {
        font-size: 28px;
        font-weight: bold;
        color: #ffffff;
    }
    </style>
""", unsafe_allow_html=True)

# 2. GENERATE MOCK DATA
@st.cache_data
def load_data():
    np.random.seed(42)
    dates = pd.date_range(start="2025-01-01", end="2025-12-31", freq="D")
    categories = ["Electronics", "Apparel", "Home & Kitchen", "Beauty & Health"]
    regions = ["North America", "Europe", "Asia-Pacific", "LATAM"]

    data = {
        "Date": np.random.choice(dates, size=2000),
        "Category": np.random.choice(categories, size=2000, p=[0.4, 0.3, 0.2, 0.1]),
        "Region": np.random.choice(regions, size=2000),
        "Sales": np.random.uniform(20, 1200, size=2000),
        "Quantity": np.random.randint(1, 10, size=2000)
    }
    df = pd.DataFrame(data)
    df["Profit"] = df["Sales"] * np.random.uniform(0.15, 0.40, size=2000)
    df["Date"] = pd.to_datetime(df["Date"])
    return df.sort_values("Date").reset_index(drop=True)

df = load_data()

# 3. SIDEBAR FILTERS
st.sidebar.header("Global Dashboard Filters")
selected_region = st.sidebar.selectbox("Select Region", ["All Regions"] + list(df["Region"].unique()))
selected_category = st.sidebar.multiselect("Select Categories", list(df["Category"].unique()), default=list(df["Category"].unique()))

# Filter Logic
filtered_df = df[df["Category"].isin(selected_category)]
if selected_region != "All Regions":
    filtered_df = filtered_df[filtered_df["Region"] == selected_region]

# 4. HEADER
st.title("📊 Executive Business Analytics Dashboard")
st.caption("Real-time corporate performance metric tracking and transaction insights.")
st.markdown("---")

# 5. KPI METRICS ROW
total_sales = filtered_df["Sales"].sum()
total_profit = filtered_df["Profit"].sum()
total_orders = filtered_df.shape[0]
avg_margin = (total_profit / total_sales) * 100 if total_sales > 0 else 0

col1, col2, col3, col4 = st.columns(4)
with col1:
    st.markdown(f'<div class="kpi-card"><div class="kpi-title">Total Revenue</div><div class="kpi-value">${total_sales:,.2f}</div></div>', unsafe_allow_html=True)
with col2:
    st.markdown(f'<div class="kpi-card" style="border-left-color: #2ecc71;"><div class="kpi-title">Net Profit</div><div class="kpi-value" style="color: #2ecc71;">${total_profit:,.2f}</div></div>', unsafe_allow_html=True)
with col3:
    st.markdown(f'<div class="kpi-card" style="border-left-color: #f1c40f;"><div class="kpi-title">Orders Fulfilled</div><div class="kpi-value" style="color: #f1c40f;">{total_orders:,}</div></div>', unsafe_allow_html=True)
with col4:
    st.markdown(f'<div class="kpi-card" style="border-left-color: #ff5722;"><div class="kpi-title">Avg Profit Margin</div><div class="kpi-value" style="color: #ff5722;">{avg_margin:.1f}%</div></div>', unsafe_allow_html=True)

st.markdown("<br>", unsafe_allow_html=True)

# 6. CHARTS GRID
chart_col1, chart_col2 = st.columns([2, 1])

with chart_col1:
    df_trend = filtered_df.groupby(pd.Grouper(key="Date", freq="W"))[["Sales"]].sum().reset_index()
    fig_trend = px.line(df_trend, x="Date", y="Sales", title="Weekly Revenue Performance Trend", template="plotly_dark")
    fig_trend.update_traces(line_color="#00adb5")
    st.plotly_chart(fig_trend, use_container_width=True)

with chart_col2:
    df_pie = filtered_df.groupby("Category")["Sales"].sum().reset_index()
    fig_pie = px.pie(df_pie, values="Sales", names="Category", title="Revenue Share by Category", hole=0.4, template="plotly_dark")
    st.plotly_chart(fig_pie, use_container_width=True)

chart_col3, chart_col4 = st.columns(2)

with chart_col3:
    df_bar = filtered_df.groupby("Category")["Quantity"].sum().reset_index()
    fig_bar = px.bar(df_bar, x="Category", y="Quantity", color="Category", title="Units Sold per Category", template="plotly_dark")
    st.plotly_chart(fig_bar, use_container_width=True)

with chart_col4:
    fig_scatter = px.scatter(filtered_df, x="Sales", y="Profit", color="Category", opacity=0.6, title="Transaction Profitability Analysis", template="plotly_dark")
    st.plotly_chart(fig_scatter, use_container_width=True)
```
